```python
import pandas as pd
import os
```

```python
#Define correct relative path to the 'data' directory for storage
data_dir = os.path.join("..", "data")
cleaned_dir = os.path.join("..", "data", "cleaned")
```

```python
#Ensure the 'cleaned' directory exists
os.makedirs(cleaned_dir, exist_ok=True)
```

```python
#Check if file exists before processing
def load_data(file_name):
    file_path = os.path.join(data_dir, file_name)
    if not os.path.exists(file_path):
        print(f"⚠️ File not found: {file_name}. Skipping.")
        return None
    return pd.read_csv(file_path)
```

### Google News Data Cleaning

```python
df_gn = load_data("GoogleNews_Data.csv")
```

```python
if df_gn is not None:
    df_gn.rename(columns={"content": "Comment", "source": "Source"}, inplace=True) #Update column names
    df_gn.drop(columns=["title"], errors="ignore", inplace=True)  #Avoid error if 'title' is missing
    df_gn = df_gn[["Comment", "Source"]]  #Keep only necessary columns
    df_gn.to_csv(os.path.join(cleaned_dir, "GoogleNews_Cleaned.csv"), index=False) #Save cleaned data
    print("✅ Google News Data cleaned and saved.")
```

### YouTube Comments Data Cleaning

```python
df_yt = load_data("YoutubeComments_Data.csv")
```

```python
if df_yt is not None:
    df_yt.rename(columns={"comment": "Comment", "source": "Source"}, inplace=True) #Update column names
    df_yt = df_yt[["Comment", "Source"]]  #Drop Timestamp column
    df_yt.to_csv(os.path.join(cleaned_dir, "YoutubeComments_Cleaned.csv"), index=False) #Save cleaned data
    print("✅ YouTube Comments Data cleaned and saved.")
```

### Reddit Data Cleaning

```python
df_rd = load_data("Reddit_Data_Comments.csv")
```

```python
if df_rd is not None:
    df_rd.rename(columns={"comment": "Comment", "source": "Source"}, inplace=True) #Update column names
    df_rd = df_rd[["Comment", "Source"]]  #Drop Timestamp column
    df_rd.to_csv(os.path.join(cleaned_dir, "RedditComments_Cleaned.csv"), index=False) #Save cleaned data
    print("✅ Reddit Data cleaned and saved.")
    
print("\n🚀 All available datasets cleaned and saved in 'data/cleaned/'.")
```
