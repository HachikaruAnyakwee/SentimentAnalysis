```python
from gnews import GNews
import pandas as pd
import os
import logging
import time
```

```python
#Configure logging
logging.basicConfig(level=logging.INFO, format='%(asctime)s - %(levelname)s - %(message)s')
```

```python
#Define search parameters
search_query = "Mark Carney"
language = "en"
country = "CA"
period = "10d"
max_results = 100
```

```python
#Set up Google News
google_news = GNews(language=language, country=country, period=period, max_results=max_results)

logging.info(f"Fetching {max_results} Google News articles for '{search_query}' in {language}-{country} over the last {period}.")

articles = []
try:
    articles = google_news.get_news(search_query)
    logging.info(f"Successfully retrieved {len(articles)} articles from Google News.")
except Exception as e:
    logging.error(f"An error occurred while fetching news from Google News: {e}")

data = []
if articles:
    for article in articles:
        try:
            data.append([article.get("published date"), article.get("title"), article.get("description"), "Google News"])
        except Exception as e:
            logging.warning(f"Error extracting data from article: {article.get('title')}. Error: {e}")
        time.sleep(1) #reasonable time lag
```

```python
#Save to CSV file
data_dir = os.path.join("..", "data")
os.makedirs(data_dir, exist_ok=True)
file_path = os.path.join(data_dir, "googlenews_data.csv")
df = pd.DataFrame(data, columns=["timestamp", "title", "content", "source"])
try:
    df.to_csv(file_path, index=False)
    logging.info(f"Successfully saved {len(df)} articles to: {file_path}")
    print(f"✅ Successfully collected {len(df)} news articles! Saved to: {file_path}")
except Exception as e:
    logging.error(f"An error occurred while saving data to CSV: {e}")
    print(f"❌ Error saving data to CSV: {e}")

if not articles:
    print("⚠️ No articles were retrieved from Google News.")
```
