```python
import pandas as pd
import os
import logging
```

```python
#Configure logging
logging.basicConfig(level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
```

```python
#Define file paths
processed_dir = os.path.join("..", "data", "processed")
final_dir = os.path.join("..", "data", "final")  #New directory for final cleaned data
os.makedirs(final_dir, exist_ok=True)  #Ensure 'final' directory exists

input_file = os.path.join(processed_dir, "cleaned_data.csv")
output_file = os.path.join(final_dir, "final_cleaned_data.csv")
```

```python
#Load the cleaned dataset
logging.info(f"Loading preprocessed dataset from: {input_file}")
df = pd.read_csv(input_file)
```

```python
#Remove missing values
initial_rows = len(df)
df_cleaned = df.dropna(subset=["text"])
missing_removed = initial_rows - len(df_cleaned)
```

```python
#Remove duplicates
initial_rows = len(df_cleaned)
df_cleaned = df_cleaned.drop_duplicates()
duplicates_removed = initial_rows - len(df_cleaned)
```

```python
#Save the final cleaned dataset
df_cleaned.to_csv(output_file, index=False)
logging.info(f"✅ Final cleaned dataset saved with {len(df_cleaned)} records to: {output_file}")
logging.info(f"Removed {missing_removed} missing values and {duplicates_removed} duplicate rows.")
```
